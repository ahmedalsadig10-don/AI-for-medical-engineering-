```python
from crewai import Agent, Task, Crew, Process

#Agent 1: The Researcher
researcher = Agent(

    role="Researcher",

    goal="Research the given topic and provide accurate information in both English and Arabic",

    backstory="You are a careful research assistant."
)
#Agnet 2:The Writer
writer = Agent(

    role="Writer",

    goal="Write a clear and well-structured article based on the research in both English and Arabic",

    backstory="You are an experienced writer."
)

#Task 1: Research
Research_task = Task(
description= "Identify key trends in the AI in healthcare for 2026 in both English and Arabic",
agent= researcher,
expected_output= "A detailed report on key trends in AI in healthcare for 2026."
)

#Task 2: writing
writer_task= Task (
description= "Compose a 3-paragrapgh blog post summarizing the key trends found in both English and Arabic",
agent= writer,
expected_output= "A 3-paragraph blog post summarizing AI in healthcare trends."
)
```

### Using Gemini API with CrewAI

To use Gemini, we can use the official `google-genai` SDK or use LangChain's Gemini integration, which CrewAI supports directly.

First, let's install the Google GenAI package:

Now, we will configure the Gemini model using CrewAI's native LLM configuration:

```python
import os
from google.colab import userdata
from crewai import LLM

# Securely load your Gemini API Key from Colab Secrets
gemini_key = userdata.get("GEMINI_API_KEY")

if not gemini_key:
    # Fallback to check for GEMINI_API_KEY if GOOGLE_API_KEY is not defined
    gemini_key = userdata.get("GEMINI_API_KEY")

if not gemini_key:
    raise ValueError("Please add your GOOGLE_API_KEY or GEMINI_API_KEY to the 🔑 Secrets panel and enable notebook access.")

# Export only one key to prevent SDK collision warnings
os.environ["GEMINI_API_KEY"] =gemini_key
if "GOOGLE_API_KEY" in os.environ:
    del os.environ["GOOGLE_API_KEY"]

# Configure the Gemini LLM for CrewAI using the correct active stable model name
gemini_llm = LLM(
    model="gemini/gemini-3.6-flash", # Updated model name to gemini-3.6-flash
    api_key=gemini_key
)

# Re-initialize agents to use Gemini
researcher.llm = gemini_llm
writer.llm = gemini_llm

print("Agents successfully configured! Model updated to gemini-3.6-flash.")
```

```python
import traceback
import os
from crewai import Crew, Process, LLM

# Configure Gemini 3.6 Flash for stability
stable_gemini_llm = LLM(
    model="gemini/gemini-3.6-flash", # Updated model name
    api_key=os.environ.get("GEMINI_API_KEY")
)

# Re-assign agents to the stable Gemini model
researcher.llm = stable_gemini_llm
writer.llm = stable_gemini_llm

try:
    # Assemble the crew with Gemini-enabled agents
    research_crew = Crew(
        agents=[researcher, writer],
        tasks=[Research_task, writer_task],
        process=Process.sequential,
        verbose=True
    )

    # Kickoff execution asynchronously within Colab's active event loop
    result = await research_crew.kickoff_async()
    print("########")
    print(result)
except Exception as e:
    print("An error occurred during kickoff:")
    traceback.print_exc()
```

### Implementing Retry Logic for API Calls

To handle temporary network glitches, rate limits, or 503 Service Unavailable errors from the LLM provider, we can configure retry logic. In CrewAI, we can leverage the underlying LiteLLM/LangChain configurations or configure a custom wrapper utilizing Python's `tenacity` library.

```python
from tenacity import retry, stop_after_attempt, wait_exponential, retry_if_exception_type
import google.genai.errors

# Define exponential backoff retry logic for CrewAI execution
@retry(
    stop=stop_after_attempt(7),                                     # Try up to 5 times
    wait=wait_exponential(multiplier=1, min=2, max=10),              # Wait 2s, 4s, 8s...
    retry=retry_if_exception_type(Exception),                       # Retry on any standard Exception
    reraise=True                                                    # Raise the last exception if all retries fail
)
async def run_crew_with_retry(crew):
    print("Starting crew execution with retry handling...")
    return await crew.kickoff_async()

# Example usage with your crew
try:
    result = await run_crew_with_retry(research_crew)
    print("######## SUCCESS ########")
    print(result)
except Exception as e:
    print(f"Execution failed after retries: {e}")
```
